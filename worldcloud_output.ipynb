```python
# Import necessary libraries
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from collections import Counter

# Ensure necessary NLTK data is downloaded
nltk.download('punkt')
nltk.download('stopwords')
```

```python
text = """
Geopolitics is the study of how geography, political power, economics, and international relations influence the behavior of countries and regions. 
It examines how factors such as territory, natural resources, population, trade routes, military capabilities, and strategic locations shape relations between states. 
Major powers often compete and cooperate to protect their national interests and expand their influence. 
International organizations such as the United Nations, World Bank, and International Monetary Fund play important roles in global affairs. 
Diplomacy, alliances, treaties, and negotiations are essential tools for managing relationships between countries and resolving conflicts. 
Economic factors such as energy supplies, trade, investment, technology, and access to critical resources can strongly influence geopolitical decisions. 
Regional organizations and partnerships also affect political and economic cooperation across different parts of the world. 
Borders, territorial disputes, migration, and access to waterways can become important sources of international tension. 
Emerging technologies, cybersecurity, artificial intelligence, and space capabilities are creating new dimensions of geopolitical competition. 
Climate change, food security, water resources, and global health can also influence international stability and cooperation. 
As the international system continues to evolve, diplomacy, economic interdependence, strategic partnerships, and national interests will remain important factors in shaping global politics and the future of international relations.
"""
print(text)
```

```python
import nltk 
nltk.download('punkt_tab')
words = word_tokenize(text.lower())  # Convert to lowercase for uniformity
print(words)
```

```python
# Remove stopwords
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word.isalpha() and word not in stop_words]
# The general syntax for a list comprehension is: [expression for item in iterable if condition]
print(filtered_words)
```

```python
word_freq=nltk.FreqDist(words)
wordcloud = WordCloud(width=800, height=400, background_color='white').generate_from_frequencies(word_freq)
```

```python
# Plotting the word cloud
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')  # Hide axes
plt.show()
```

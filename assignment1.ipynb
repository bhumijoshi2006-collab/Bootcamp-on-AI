```python
import pandas as pd
import numpy as np


random_numbers = np.random.randint(1, 100, size=15).tolist()
print("Original list:", random_numbers)


s = pd.Series(random_numbers)


even_numbers = s[s % 2 == 0]
print("Even numbers:", even_numbers.tolist())
```

```python
import pandas as pd


text = "Pandas makes data analysis easy"
print("Original string:", text)

reversed_string = text[::-1]
print("Reversed string:", reversed_string)


words_series = pd.Series(text.split())
reversed_words = ' '.join(words_series[::-1])
print("Reversed words:", reversed_words)
```

```python
import cv2
import numpy as np
import matplotlib as plt
```

```python
img=cv2.imread("8iIyRh.jpg")
print(type(img))
```

```python
print(img.shape)
```

```python
import matplotlib.pyplot as plt

plt.imshow(img)
plt.axis("off")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Resize image to 200 x 200
img_resize = cv2.resize(img, (200, 200))

# Convert BGR to RGB
img_resize_rgb = cv2.cvtColor(img_resize, cv2.COLOR_BGR2RGB)

# Display resized image
plt.imshow(img_resize_rgb)
plt.axis("off")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Flip image horizontally
img_flip = cv2.flip(img, 1)

# Convert BGR to RGB for displaying
img_flip_rgb = cv2.cvtColor(img_flip, cv2.COLOR_BGR2RGB)

# Display image
plt.imshow(img_flip_rgb)
plt.axis("off")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Flip image vertically
img_flip_1 = cv2.flip(img, 0)

# Convert BGR to RGB
img_flip_1_rgb = cv2.cvtColor(img_flip_1, cv2.COLOR_BGR2RGB)

# Display image
plt.imshow(img_flip_1_rgb)
plt.axis("off")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Convert image from BGR to Grayscale
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Display grayscale image
plt.imshow(img_gray, cmap="gray")
plt.axis("off")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Crop the image
img_crop = img[100:300, 200:400]

# Display cropped image
plt.imshow(cv2.cvtColor(img_crop, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```
